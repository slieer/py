```python
# 线性代数第二页
# 解线性方程组，根据增广矩阵求解
# x1 - 2x2 = -1
# -x + 3x2 = 3
import numpy as np

# fmt: off

# 增广矩阵(Augmented Matrix)
augmented = np.array([[1, -2, -1], 
                      [-1, 3, 3]])

augmented = np.array([[1, 5, 7], 
                      [1, -2, -2]])

augmented = np.array([[1, -5, 1], 
                      [3, -7, 5]])
# fmt: on

A = augmented[:, :-1]  # 提取系数矩阵
b = augmented[:, -1]  # 提取常数项向量
print(f"系数矩阵:\n{A}")
print(f"常数项向量: \n{b}")
x = np.linalg.solve(A, b)

print(f"线性方程组的解: {x}")
```

```python
# 解线性方程组(System Of Linear Equations)，根据系数矩阵A和常数向量b求解
# 2x + y = 5
# x - 3y = -1
# fmt: off
A = np.array([[2, 1], 
              [1, -3]
            ])
b = np.array([5, -1])
# fmt: on

# 使用 np.linalg.solve 直接求解
# np.linalg.solve 只能处理唯一解的情况。
# 当矩阵奇异时，你需要改用其他方法来判断是“无解”还是“无穷多解”，并求出通解或特解。
x = np.linalg.solve(A, b)
print(f"方程组的解: {x}")  # 输出: [2. 1.]，即 x=2, y=1
```

```python
# 判断增广矩阵(Augmented Matrix)是否相容

# fmt: off
# LinAlgError: Singular matrix, 不能用solve求解，因为矩阵奇异。
augmented = np.array([[0, 1, -4, 8], 
                      [2, -3, 2, 1], 
                      [4, -8, 12, 1]])

# 练习题第一章第2题
augmented = np.array([[1, 5, 2, -6],
                      [0, 4,-7, 2], 
                      [0, 0, 5, 0]                      
                      ])

augmented = np.array([[1, 2, 1, 4],
                      [0, 1, -1, 1], 
                      [1, 3, 0, 0]                      
                      ])
# fmt: on
# 系数矩阵(Coefficient Matrix)
coM = augmented[:, :-1].astype(float)
# 未知数个数（Number of unknowns）
n_vars = coM.shape[1]
# 常数项向量(Constant Vector)
constV = augmented[:, -1].astype(float)

# 秩(Rank)：指矩阵中线性无关的行或列的最大数目）
# 系数矩阵的秩
rank_A = np.linalg.matrix_rank(coM)
# 增广矩阵的秩
rank_Ab = np.linalg.matrix_rank(augmented.astype(float))

print(f"系数矩阵 A 的秩: {rank_A}")
print(f"增广矩阵 [A|b] 的秩: {rank_Ab}")
print(f"未知数个数: {n_vars}")

# 系数矩阵，增广矩阵，未知数个数相等，唯一解
if len({rank_A, rank_Ab, n_vars}) == 1:
    print("线性方程组有唯一解")
    x = np.linalg.solve(coM, constV)
    print(f"方程组的解为: {x}")
elif rank_A == rank_Ab < n_vars:
    print("线性方程组有无穷多解")
    x = np.linalg.lstsq(coM, constV, rcond=None)
    print(f"方程组的解为: {x}")
elif rank_A < rank_Ab:
    print("线性方程组无解")
```

```python
# 验证向量是不是线性方程组的一个解


def is_solution(coA, constA, x_candidate):
    """
    判断向量 x_candidate 是否是线性方程组 Ax = b 的解。
    """
    # 1. 计算 Ax
    Ax = np.dot(coA, x_candidate)
    print(Ax)
    # print(f"coA = {coA}, x_candidate = {x_candidate}, Ax = {Ax}")
    # 2. 验证 Ax 是否等于 b
    # 使用 allclose 来处理浮点数的精度问题
    return np.allclose(Ax, constA)


# fmt: off
#  5x1 -  x2 + 2x3 = 7
# -2x1 + 6x2 + 9x3 = 0
# -7x1 + 5x2 - 3x3 = -7
A = np.array([[5, -1, 2, 7], 
              [-2, 6, 9, 0], 
              [-7, 5, -3, -7]
            ])
# fmt: on

x_candidate = np.array([3, 4, -2])  # 一个候选解

constA = A[:, -1]  # 提取常数项向量
coA = A[:, :-1]  # 提取系数矩阵

is_one_solution = is_solution(coA, constA, x_candidate)
print(f"向量 {x_candidate} 是否是线性方程组的解: {is_one_solution}")

# 增广矩阵的秩
rank_A = np.linalg.matrix_rank(A)
# 系数矩阵的秩
rank_coA = np.linalg.matrix_rank(coA)
# 变量个数
n_vars = coA.shape[1]
print(
    f"增广矩阵augmented的秩: {rank_A}, 系数矩阵 coA 的秩: {rank_coA}, 变量个数: {n_vars}"
)
if rank_A == rank_coA == n_vars:
    print("线性方程组有唯一解")
elif rank_A == rank_coA:
    print("线性方程组有无穷多解")
elif rank_A != rank_coA:
    print("线性方程组无解")
```

```python
# 判断是否有唯一解 无穷多解 无解

# fmt: off
# 无穷多解 (Underdetermined)
augmented_matrix = np.array([[1, 1, 2],
                             [2, 2, 4]])

augmented_matrix = np.array([[1, 1, 2],
                             [2, 5, 7]])

# fmt: on

A = augmented_matrix[:, :-1].astype(float)
b = augmented_matrix[:, -1].astype(float)

# lstsq 返回一个元组 (x, residuals, rank, singular_values)，其中：
# x 是最小二乘解（如果存在唯一解，则为该解；如果无穷多解，则为其中一个特解）。
# residuals 是残差向量，即 A x - b。
# rank 是矩阵 A 的秩。
# singular_values 是矩阵 A 的奇异值。
import sympy as sp

x_special, residuals, rank, s = np.linalg.lstsq(A, b, rcond=None)

print(f"x_special= {x_special}, residuals= {residuals}, rank= {rank}, s= {s}")
print(
    {
        "特解 (最小范数)": x_special,
        "自由变量个数": n_vars - rank_A,
        "残差": residuals,
        "秩": rank_A,
    },
)
```

```python
import sympy
from sympy import Matrix, symbols, Eq, solve

# 方程组:
# x1 - 2x2 + x3 = 0
#      2x2 - 8x3 = 8
# 5x1       -5x3 = 10

# 创建对应的增广矩阵
# fmt: off
augmented = [
    [1, -2, 1, 0],
    [0, 2, -8, 8],
    [5, 0, -5, 10]
]
# fmt: on

augmented_matrix = Matrix(augmented)
# 进行行化简至最简行阶梯形 (RREF)
rref_matrix, pivot_columns = augmented_matrix.rref()

print("增广矩阵的RREF形式:")
sympy.pprint(rref_matrix)

"""
[[1, 0, 0, 1], 
 [0, 1, 0, 0], 
 [0, 0, 1, -1]
]
"""
r = rref_matrix[:, -1]
print(f"方程组解: {r}")

x1 = rref_matrix[0, -1]
x2 = rref_matrix[1, -1]
x3 = rref_matrix[2, -1]
print(f"方程组解: x1 = {x1}, x2 = {x2}, x3 = {x3}")
```

```python
import sympy as sp

# 1. 定义符号
h, k = sp.symbols("h k")

# 2. 构建增广矩阵
# fmt: off
M = sp.Matrix([[2, -1, h], 
               [-6, 3, k]])
# fmt on

print("原始增广矩阵：")
sp.pprint(M)

# 3. 手动进行行变换 (避免自动 rref 的“激进”化简)
# 目标：消去第二行的 -6
# 操作：第2行 = 第2行 + 3 * 第1行
# 在 sympy 中，row_op 是原地操作，我们需要先复制一份或者一步步来
M_transformed = M.copy()
M_transformed.row_op(1, lambda v, j: v + 3 * M_transformed[0, j])

print("\n手动消元后的矩阵 (这才是我们要看的)：")
sp.pprint(M_transformed)

# 4. 提取相容性条件
# 现在我们看到第二行是 [0, 0, 3*h + k]
# 为了相容，必须满足 3*h + k = 0
condition_expr = M_transformed[1, 2]

print("\n--- 结论 ---")
print(f"相容性限制条件: {condition_expr} = 0")

# 5. 求解 k
solution = sp.solve(condition_expr, k)

if solution:
    print(f"解得: k = {solution[0]}")
else:
    print("无解或条件恒成立")
```

```python
# 练习题第一章第4题 --python 手算方式

from sympy import symbols, Matrix, solve

h, k = symbols("h k")
# fmt: off
aug = Matrix([[2, -1, h], 
              [-6, 3, k]])
# fmt: on

# 执行行变换：R2 = R2 + 3*R1
# 这一步比 rref 更可控，因为 rref 可能会除以符号
row1 = aug.row(0)
row2 = aug.row(1) + 3 * aug.row(0)

# 新的第二行是 [0, 0, k + 3*h]
# 要有解，必须 k + 3*h = 0
equation = row2[2]
print(f"row2[2]: " + str(equation))

print(f"约束方程: {equation} = 0")
result = solve(equation, k)
print(f"k = {result[0]}")
```
